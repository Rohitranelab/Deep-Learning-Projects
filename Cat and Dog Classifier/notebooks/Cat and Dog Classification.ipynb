```python
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
```

```python
Image_Width = 128
Image_Height = 128
Image_Size = (Image_Width, Image_Height)
Image_Channels = 3
```

```python
cats_path = '/content/cat_and_dog/training_set/training_set/cats'
dogs_path = '/content/cat_and_dog/training_set/training_set/dogs'
```

```python
import os
import pandas as pd

train_path = "/content/cat_and_dog/training_set/training_set"

filenames = []
categories = []

for file in os.listdir(os.path.join(train_path, "cats")):
    filenames.append("cats/" + file)
    categories.append("cat")

for file in os.listdir(os.path.join(train_path, "dogs")):
    filenames.append("dogs/" + file)
    categories.append("dog")

train_df = pd.DataFrame({
    "filename": filenames,
    "category": categories
})
```

```python
test_path = "/content/cat_and_dog/test_set/test_set"

filenames = []
categories = []

for file in os.listdir(os.path.join(test_path, "cats")):
    filenames.append("cats/" + file)
    categories.append("cat")

for file in os.listdir(os.path.join(test_path, "dogs")):
    filenames.append("dogs/" + file)
    categories.append("dog")

validate_df = pd.DataFrame({
    "filename": filenames,
    "category": categories
})
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Flatten, Conv2D, MaxPooling2D, BatchNormalization
```

```python
model = Sequential()

model.add(Conv2D(32, (3, 3), activation = 'relu', input_shape = (Image_Width, Image_Height, Image_Channels)))
model.add(BatchNormalization())
model.add(MaxPooling2D(pool_size = (2, 2)))
model.add(Dropout(0.2))

model.add(Conv2D(64, (3, 3), activation = 'relu'))
model.add(BatchNormalization())
model.add(MaxPooling2D(pool_size = (2, 2)))
model.add(Dropout(0.2))

model.add(Conv2D(128, (3, 3), activation = 'relu'))
model.add(BatchNormalization())
model.add(MaxPooling2D(pool_size = (2, 2)))
model.add(Dropout(0.2))

model.add(Flatten())
model.add(Dense(512, activation = 'relu'))
model.add(BatchNormalization())
model.add(Dropout(0.2))
model.add(Dense(2, activation = 'softmax'))
```

```python
model.summary()
```

```python
model.compile(optimizer = 'adam', loss = 'categorical_crossentropy', metrics = ['accuracy'])
```

```python
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
```

```python
earlystop = EarlyStopping(patience = 10)
learning_rate_reduction = ReduceLROnPlateau(monitor = 'val_loss', patience = 2, verbose = 1, factor = 0.5, min_lr = 0.00001)
callbacks = [earlystop, learning_rate_reduction]
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator, load_img
from tensorflow.keras.utils import to_categorical
```

```python
train_datagen = ImageDataGenerator(rotation_range = 15, rescale = 1./255,
                                   shear_range = 0.1, zoom_range = 0.2,
                                   horizontal_flip = True, width_shift_range = 0.1,
                                   height_shift_range = 0.1)

validation_datagen = ImageDataGenerator(rescale = 1./255)

test_datagen = ImageDataGenerator(rotation_range=15, rescale=1./255,
                                shear_range=0.1, zoom_range=0.2,
                                horizontal_flip=True, width_shift_range=0.1, height_shift_range=0.1)
```

```python
train_generator = train_datagen.flow_from_dataframe(
    dataframe=train_df,
    directory=train_path,
    x_col="filename",
    y_col="category",
    target_size=Image_Size,
    class_mode="categorical",
    batch_size=15,
    shuffle=True
)

validation_generator = validation_datagen.flow_from_dataframe(
    dataframe=validate_df,
    directory=test_path,
    x_col="filename",
    y_col="category",
    target_size=Image_Size,
    class_mode="categorical",
    batch_size=15,
    shuffle=False
)

test_generator = test_datagen.flow_from_dataframe(
    dataframe=validate_df,
    directory=test_path,
    x_col="filename",
    y_col="category",
    target_size=Image_Size,
    class_mode="categorical",
    batch_size=15,
    shuffle=False
)
```

```python
history = model.fit(
    train_generator,
    epochs = 30,
    validation_data = validation_generator,
    callbacks = callbacks
)
```

```python
model.save('cat_and_dog.keras')
```
